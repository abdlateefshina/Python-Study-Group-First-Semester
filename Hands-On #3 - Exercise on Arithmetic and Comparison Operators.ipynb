# Question 1
## Employee Payroll & Bonus Eligibility

```python
# Below is the Payroll & Bonus Eligibility for an employee.

basic_salary = 180000
performance_bonus = 45000
total_salary = basic_salary + performance_bonus
print ("Total Salary:", total_salary)

# Determining whether the employee is eligible.
premium_benefits_eligibility = 220000
print ("Eligible for Premium Benefits:", total_salary >= premium_benefits_eligibility)
```

# Question 2
## Student Performance Evaluation

```python
# Below is the Performance Evaluation for a student.

python = 72
sql = 65
excel = 80
power_bi = 75
total_score = python + sql + excel + power_bi
print ("Total Score:", total_score)

# Finding the Average Score.
average_score = total_score / 4
print ("Average Score:", average_score)

# Determinig whether the student passed.
pass_mark = 70
print ("Passed:", average_score >= pass_mark)
```

# Question 3
## Inventory Management System

```python
# Below is the Inventory Management System for a warehouse

initial_stock = 250
sold_stock = 85
remaining_stock = initial_stock - sold_stock
print ("Remaining Stock:", remaining_stock)

# Determining whether the warehouse should reorder inventory.
reorder_limit = 200
print ("Reorder Needed:", remaining_stock < reorder_limit)
```

# Question 4
## Banking Withdrawal Checker

```python
# Below is a Withdrawal Checker for a bank.

account_balance = 520000
withdrawal_amount = 150000
remaining_balance = account_balance - withdrawal_amount
print ("Remaining Balance:", remaining_balance)

# Determining whether withdrawal can be approved.
print ("Withdrawal Approved:", account_balance >= withdrawal_amount)
```

# Bonus Challenge
## Smart Shopping Discount Calculator

```python
# Below is a Smart Discount Calculator for a customer.

laptop_bag = 18500
wireless_mouse = 12000
keyboard = 24500
usb_drive = 5000
total_purchase = laptop_bag + wireless_mouse + keyboard + usb_drive
print ("Total Purchase:", total_purchase)

# Check whether the customer is qualified for discount.
minimum_purchase = 50000
print ("Discount Qualified:", total_purchase >= minimum_purchase)

# Calculate the discount amount.
discount_rate = 10 / 100
discount_amount = discount_rate * total_purchase
print ("Discount Amount:", discount_amount)

# Calculate the final amount to be paid by the customer
final_amount = total_purchase - discount_amount
print ("Final Amount:", final_amount)
```

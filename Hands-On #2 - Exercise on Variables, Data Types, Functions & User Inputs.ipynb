# Mini Project: Student Information Card

```python
# Gave all the student details a user input function
name = input ("What is your name?")
age = input ("How old are you")
course = input ("Which course are you applying for?")
country = input ("Which country are you from?")
# Then I displayed the Student Information Card using a string and its corresponding variable
print ("------------------------------------------------------------------------")
print ("------------------------------------------------------------------------")
print ("STUDENT INFORMATION CARD")
print ("------------------------------------------------------------------------")
print ("------------------------------------------------------------------------")

print ("Name:", name)
print ("Age:", age)
print ("Course:", course)
print ("Country:", country)

print ("------------------------------------------------------------------------")
print ("DATA TYPES")
print ("------------------------------------------------------------------------")

print ("Name:", type (name))
print ("Age:", type (age))
print ("Course:", type (course))
print ("Country:", type (country))
```

## Changing the user's age into an integer

```python
# Gave all the student details a user input function
name = input ("What is your name?")
# Changing the user's age into an integer
age = int (input ("How old are you"))
course = input ("Which course are you applying for?")
country = input ("Which country are you from?")
# Then I displayed the Student Information Card using a string and its corresponding variable
print ("------------------------------------------------------------------------")
print ("------------------------------------------------------------------------")
print ("STUDENT INFORMATION CARD")
print ("------------------------------------------------------------------------")
print ("------------------------------------------------------------------------")

print ("Name:", name)
print ("Age:", age)
print ("Course:", course)
print ("Country:", country)

print ("------------------------------------------------------------------------")
print ("DATA TYPES")
print ("------------------------------------------------------------------------")

print ("Name:", type (name))
print ("Age:", type (age))
print ("Course:", type (course))
print ("Country:", type (country))
```

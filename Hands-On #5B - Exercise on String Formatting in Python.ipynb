# Exercise 2
## Payroll System

```python
# Creating the variables.
employee_name = "David Musa"
employee_department = "Finance"
employee_basic_salary = 185000.00
employee_bonus = 35000.00
employee_tax_rate = 7.5
employee_gross_salary = employee_basic_salary + employee_bonus
employee_tax_amount = (employee_tax_rate / 100) * employee_gross_salary

# Displaying the payroll receipt.
employee_payroll_receipt = f"""
======================================
            PAYROLL SUMMARY
======================================

Employee: {employee_name}
Department: {employee_department}
Basic Salary: {employee_basic_salary}
Bonus: {employee_bonus}
Gross Salary: {employee_basic_salary + employee_bonus}
Tax ({employee_tax_rate}%): {(employee_tax_rate / 100) * employee_gross_salary}
Net Salary: {employee_gross_salary - employee_tax_amount}

======================================
"""

print (employee_payroll_receipt)
```

# Exercise 3
## E-commerce Order Receipt

```python
# Creating the variables.
customer_name = "Amina Bello"
product_name = "Wireless Headphones"
product_quantity = 3
product_unit_price = 27500
total_cost = product_unit_price * product_quantity

# Displaying the order receipt.
customer_order_receipt = f"""
======================================
        ONLINE STORE RECEIPT
======================================

Customer: {customer_name}
Product: {product_name}
Quantity: {product_quantity}
Unit Price: {product_unit_price}
Total Cost: {total_cost}

Thank you for shopping with us!
======================================
"""

print (customer_order_receipt)
```

### Bonus Challenge

```python
# Creating the variables.
customer_name = "Amina Bello"
product_name = "Wireless Headphones"
product_quantity = 3
product_unit_price = 27500
discount_rate = 10
total_cost = product_unit_price * product_quantity
discount_amount = (discount_rate / 100) * total_cost

# Displaying the order receipt.
customer_order_receipt = f"""
======================================
        ONLINE STORE RECEIPT
======================================

Customer: {customer_name}
Product: {product_name}
Quantity: {product_quantity}
Unit Price: {product_unit_price}
Total Cost: {total_cost}
Discount: {discount_amount}
Final Amount: {total_cost - discount_amount}

Thank you for shopping with us!
======================================
"""

print (customer_order_receipt)
```

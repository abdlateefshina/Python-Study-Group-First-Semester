# Hands-on 1
## Student Score Tracker

```python
# Assigning the variables.
score = int (input ("Initial Score:"))
class_assignment = 15
late_submission = 5
special_bonus = 2
```

```python
# Updating the student's score.
score += class_assignment
score -= late_submission
score *= special_bonus

# Displaying the final score.
print ("Final Score:", score)
```

# Hands-on 2
## Digital Wallet Simulator

```python
# Assigning the variable.
balance = 25000
deposit = int (input ("Amount Deposited:"))
withdrawal = int (input ("Amount Withdrew:"))
savings_accounts = 3
```

```python
# Updating the balance.
balance += deposit
balance -= withdrawal
balance /= savings_accounts

# Displaying the current balance.
print ("Current Balance:", balance)
print (type (balance))
```

# Hands-on 3
## Inventory Management System

```python
# Assigning the variables.
stock = int (input("Initial Stock:"))
items_sold = int (input ("Items Sold:"))
items_received = int (input ("Items Received:"))
items_per_carton = int (input ("Items per Carton:"))
```

```python
# Updating the items.

# After sales
stock -= items_sold
print ("Remaining stock after sales:", stock)

# After restocking
stock += items_received
print ("Updated stock after restocking:", stock)

# After packing into cartons
stock //= items_per_carton
print ("Number of full cartons:", stock)

items_after_restocking = 197
unpacked_items = items_after_restocking % items_per_carton
print ("Remaining unpacked items:", unpacked_items)
```
